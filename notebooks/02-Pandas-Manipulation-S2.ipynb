## 1. Introducing DataFrames

`pandas` is a Python library designed for data manipulation and analysis.
Its central data structure is the **DataFrame**, a two-dimensional table
organized into rows and columns.

Some of the first methods we use when discovering a dataset are:

- `head()` — display the first rows
- `info()` — inspect columns, data types, and missing values
- `describe()` — compute descriptive statistics

On crée un petit DataFrame à partir d'un dictionnaire. Chaque clé devient une colonne et chaque liste donne les valeurs de cette colonne.

```python
import pandas as pd

data = {
    "Name": ["Fido", "Buddy", "Rex"],
    "Age": [4, 5, 6]
}

df = pd.DataFrame(data)

df.head()
```

On obtient un tableau de 3 lignes et 2 colonnes, avec un index de 0 à 2 ajouté automatiquement par pandas.

### From NumPy to pandas
A NumPy array gives you numerical operations; pandas adds **column names, row labels and heterogeneous column types**.
A `Series` is one labelled column. A `DataFrame` is a table containing several columns.
Run the examples, then complete the exercises in the empty cells. Explain what each result means, not only which method you used.

On compare `df["Age"]` et `df[["Age"]]`. Avec un seul crochet on obtient une colonne seule (une Series), avec deux crochets on obtient un tableau d'une colonne (un DataFrame).

```python
print(type(df["Age"]))
print(type(df[["Age"]]))
df[["Name"]]
```

Les types affichés confirment la différence, `Series` d'un côté et `DataFrame` de l'autre. Le dernier affichage montre la colonne `Name` sous forme de tableau.

## 3. Selecting, filtering and sorting
Select a column with `table["column"]`; select several columns with a list of names.
A comparison creates a Boolean mask, just as with NumPy. `.loc[rows, columns]` lets you choose both rows and columns.

On sélectionne seulement deux colonnes en donnant une liste de noms.

```python
homelessness[["state", "individuals"]].head()
```

Le résultat ne garde que l'État et le nombre de personnes seules sans domicile.

On crée un masque qui vaut vrai pour les États de la région Mountain, puis `.loc` garde ces lignes et seulement trois colonnes.

```python
mask = homelessness["region"] == "Mountain"
homelessness.loc[mask, ["state", "individuals", "state_pop"]]
```

On obtient les 8 États de la région Mountain.

Combine conditions with `&` (and), `|` (or), and `~` (not). Put each comparison in parentheses; do not use Python's `and` between Series.

On combine deux conditions avec `&`, chacune entre parenthèses. On garde les États de la région Mountain qui ont plus de 1 000 personnes seules sans domicile.

```python
homelessness.loc[
    (homelessness["region"] == "Mountain") & (homelessness["individuals"] > 1000),
    ["state", "individuals"]
]
```

Il reste 6 États. Montana et Wyoming disparaissent car ils sont sous le seuil de 1 000.

On trie le tableau par nombre décroissant de personnes seules sans domicile et on garde les cinq premiers États.

```python
homelessness.sort_values("individuals", ascending=False)[["state", "individuals"]].head(5)
```

La Californie arrive très largement en tête, devant New York, la Floride, le Texas et l'État de Washington.

`head(5)` keeps the first five rows of the current order. `.iloc` selects by position; `.loc` selects by labels or Boolean masks. Neither sorting nor selecting changes the original table unless you assign the result.

On sélectionne par position avec `.iloc`, ici les trois premières lignes et les deux premières colonnes.

```python
homelessness.iloc[:3, :2]
```

On obtient les régions et les noms des trois premiers États du fichier.

## 4. Creating meaningful indicators
Vectorized calculations work on entire columns. There is no need to loop over rows.
Here is a rate for **individuals only**; people in families are not included in this indicator.

On crée un taux pour 10 000 habitants en divisant le nombre de personnes seules sans domicile par la population de l'État, puis en multipliant par 10 000. Le calcul se fait sur toute la colonne d'un coup.

```python
homelessness["individuals_per_10k"] = homelessness["individuals"] / homelessness["state_pop"] * 10_000
homelessness[["state", "individuals_per_10k"]].head().round(2)
```

Un taux permet de comparer des États de tailles différentes. La Californie a par exemple environ 27,6 personnes seules sans domicile pour 10 000 habitants.

## 5. Summarising data with pandas
A summary reduces many observations to a few numbers. Choose the variable and the question first.
`count()` counts non-missing values; `size()` counts rows. `value_counts()` counts occurrences of each category.

On calcule plusieurs résumés de la même colonne en une seule fois avec `agg()`.

```python
homelessness["individuals"].agg(["sum", "mean", "median", "max"])
```

Il y a 368 515 personnes seules sans domicile au total. La médiane (3 082) est bien plus basse que la moyenne, à cause des très gros États.

On compte le nombre d'États dans chaque région avec `value_counts()`.

```python
homelessness["region"].value_counts()
```

La région South Atlantic contient le plus d'États (9) et Mid-Atlantic le moins (3). Les régions ne sont donc pas de même taille.

### From a single summary to one summary per group
`groupby()` follows three steps: split rows into groups, apply a calculation, and combine the results.
The following example sums counts within each region. It does not average state rates.

On regroupe les États par région avec `groupby()` et on additionne les comptages et les populations de chaque région.

```python
regional = homelessness.groupby("region")[["individuals", "family_members", "state_pop"]].sum()
regional
```

La région Pacific a de loin le plus de personnes seules sans domicile (142 136). Les noms des régions sont passés dans l'index.

The group labels now form the index. `reset_index()` turns those labels back into a regular column.

On remet les noms des régions dans une colonne normale avec `reset_index()`, pour pouvoir les utiliser comme les autres colonnes.

```python
regional = regional.reset_index()
regional.head()
```

La colonne `region` est revenue, et l'index est de nouveau numéroté à partir de 0.

### Your turn — Compare regions fairly
Using `regional`, create the total homelessness count and the total rate per 10,000 inhabitants for each region.
Sort by rate, highest first. Explain why you divide the **sum of counts by the sum of populations**, rather than taking the simple average of state rates.

Notre réponse. Pour chaque région, on additionne les personnes seules et les membres de familles, puis on divise ce total par la population totale de la région et on multiplie par 10 000. On trie ensuite par taux décroissant.

```python
regional["total_homeless"] = regional["individuals"] + regional["family_members"]
regional["total_per_10k"] = regional["total_homeless"] / regional["state_pop"] * 10_000

regional_rates = regional.sort_values("total_per_10k", ascending=False)[
    ["region", "total_homeless", "state_pop", "total_per_10k"]
]
regional_rates.round(2)
```

La région Pacific a le taux le plus élevé (environ 32,9 pour 10 000 habitants), devant Mid-Atlantic et New England.

On divise la somme des comptages par la somme des populations, car c'est le vrai taux de la région, où chaque habitant compte autant. La moyenne simple des taux des États donnerait le même poids à un petit État qu'à un très grand. Par exemple, le district de Columbia, très petit mais avec un taux d'environ 98, ferait monter la région South Atlantic à la troisième place, alors que son vrai taux la place cinquième.

Chaque ligne décrit les ventes d'un rayon d'un magasin pour une semaine, avec des informations de contexte comme la température et le carburant.

#### Étape 1, inspection du tableau
Avant de répondre au responsable, on regarde la taille du tableau, le nombre de magasins et le nombre de ventes négatives. L'énoncé demande de garder les ventes négatives et de signaler leur présence, sans les supprimer.

```python
print(f"Le tableau contient {sales.shape[0]} lignes et {sales.shape[1]} colonnes.")
print(f"Nombre de magasins différents {sales['store'].nunique()}")
print(f"Nombre d'observations avec des ventes négatives {(sales['weekly_sales'] < 0).sum()}")
```

Le tableau contient 10 774 lignes pour 12 magasins, et 108 observations ont des ventes négatives. Elles sont gardées telles quelles.

#### Étape 2, résumé par type de magasin
Pour chaque type de magasin, on calcule en une seule fois le total des ventes, la moyenne par observation et le nombre d'observations avec `agg()`.

```python
type_summary = sales.groupby("type")["weekly_sales"].agg(
    total_sales="sum",
    mean_sales="mean",
    observations="size",
)
type_summary.round(2)
```

Le type A a un total environ dix fois plus grand que le type B, mais il a aussi environ onze fois plus d'observations (9 872 contre 902).

#### Étape 3, le même type est-il en tête sur le total et sur la moyenne ?
On cherche le type en tête pour chaque indicateur avec `idxmax()`. On compte aussi le nombre de magasins de chaque type, qui aide à expliquer l'écart.

```python
leader_total = type_summary["total_sales"].idxmax()
leader_mean = type_summary["mean_sales"].idxmax()
print(f"Le type {leader_total} a le plus gros total, le type {leader_mean} a la plus grosse moyenne par observation.")
display(sales.groupby("type")["store"].nunique().rename("stores"))
```

Ce n'est pas le même type. Le type A est en tête sur le total, le type B sur la moyenne par observation. Le type A regroupe 11 magasins de l'échantillon et le type B un seul, ce qui explique en grande partie l'écart des totaux.

#### Étape 4, les cinq magasins au plus gros total
On fait le même calcul par magasin, puis on trie par total décroissant et on affiche les cinq premiers, avec leur nombre d'observations.

```python
store_summary = (
    sales.groupby("store")["weekly_sales"]
    .agg(total_sales="sum", observations="size")
    .sort_values("total_sales", ascending=False)
)
store_summary.head(5).round(2)
```

Le magasin 14 a le plus gros total. Les nombres d'observations des cinq premiers sont proches (entre 885 et 913), donc leurs totaux sont comparables entre eux.

#### Étape 6, export du résumé par type
On enregistre le résumé par type dans `sales_by_type.csv`. `reset_index()` remet le type dans une colonne normale et `index=False` évite d'écrire l'index de pandas dans le fichier.

```python
type_summary.reset_index().to_csv("sales_by_type.csv", index=False)
print("Le fichier sales_by_type.csv est enregistré dans le dossier du notebook.")
```

Le fichier est créé dans le dossier `notebooks`, à côté du notebook.

**Group interpretation:**

Le type A a de loin le plus gros total de ventes (environ 233,7 millions de dollars contre 23,2 millions pour le type B), mais c'est surtout parce que l'échantillon contient 9 872 observations du type A, réparties sur 11 magasins, contre seulement 902 pour le type B, qui n'a qu'un magasin. Par observation, c'est au contraire le type B qui vend le plus en moyenne (environ 25 697 dollars contre 23 675 pour le type A). Une moyenne par observation n'est pas une moyenne par magasin, car un magasin qui a plus de lignes dans l'échantillon pèse davantage, et ce n'est pas une preuve de rentabilité puisqu'on ne connaît ni les coûts ni les marges. L'affirmation selon laquelle le type au plus gros total serait le plus performant n'est donc pas démontrée par cet échantillon, qui contient en plus 108 ventes hebdomadaires négatives gardées telles quelles.
