### 2.1 Lists and indexing

A Python list stores an ordered collection of values.

Python uses **zero-based indexing**: the first element has index `0`.

On range des notes dans une liste puis on accède à certaines valeurs par leur position. On utilise `0` pour le premier élément, `-1` pour le dernier et `[:3]` pour les trois premiers.

````python
scores = [12, 15, 9, 18, 14]

print("All scores:", scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("First three scores:", scores[:3])
````

La première note est 12 et la dernière 14. L'indexation commence bien à 0 en Python, et `[:3]` s'arrête avant la position 3.

### 2.2 Dictionaries

A Python dictionary stores data as **key-value pairs**.

Dictionaries are useful when data have meaningful labels rather than numerical positions.

On décrit une étudiante avec un dictionnaire, où chaque information est associée à une clé lisible comme `name` ou `grade`. On récupère ensuite une valeur grâce à sa clé.

````python
student = {
    "name": "Alice",
    "program": "SBI",
    "grade": 15.5,
    "passed": True
}

print(student)
print("Name:", student["name"])
print("Grade:", student["grade"])
````

Le dictionnaire complet s'affiche, puis le nom et la note obtenus avec leur clé. On n'a pas besoin de connaître la position d'une information pour la retrouver.

On modifie une valeur existante (la note) et on ajoute une nouvelle clé (l'année), simplement en leur affectant une valeur.

````python
student["grade"] = 16
student["year"] = "M1"

print(student)
````

La note passe à 16 et la clé `year` apparaît à la fin du dictionnaire.

### 2.3 Conditions and loops

Conditions allow a program to make decisions.

Loops allow us to repeat an operation over several values.

On parcourt la liste des notes avec une boucle `for` et on teste chaque note avec `if`. Une note supérieure ou égale à 10 est validée, sinon elle ne l'est pas.

````python
scores = [12, 15, 9, 18, 14]

for score in scores:
    if score >= 10:
        print(score, "→ Pass")
    else:
        print(score, "→ Fail")
````

Chaque note est suivie de son résultat. Seule la note 9 est inférieure à 10, donc c'est la seule marquée « Fail ».

### 2.4 Functions

A function groups reusable instructions.

Functions help us avoid repeating code and make data-processing workflows easier to understand.

On regroupe la règle de validation dans une fonction `classify_score`. On peut ainsi la réutiliser sur n'importe quelle note sans réécrire le test.

````python
def classify_score(score):
    if score >= 10:
        return "Pass"
    else:
        return "Fail"

print(classify_score(15))
print(classify_score(8))
````

La fonction renvoie « Pass » pour 15 et « Fail » pour 8, ce qui correspond à la règle du seuil de 10.

Notre réponse au mini-challenge, en Python simple. On calcule d'abord le nombre de commandes avec `len()`, le total avec `sum()` puis la moyenne. On classe ensuite chaque montant avec la fonction `classify_sales`, on construit la liste des ventes élevées avec une boucle, et on regroupe les indicateurs dans le dictionnaire `report`.

````python
sales = [120, 95, 140, 180, 75, 160]

number_of_orders = len(sales)
total_sales = sum(sales)
mean_sales = total_sales / number_of_orders

print("Nombre de commandes :", number_of_orders)
print("Total :", total_sales)
print("Moyenne :", mean_sales)

def classify_sales(value):
    if value >= 150:
        return "High"
    else:
        return "Standard"

for value in sales:
    print(value, "→", classify_sales(value))

high_sales = []

for value in sales:
    if value >= 150:
        high_sales.append(value)

print("Ventes High :", high_sales)

report = {
    "orders": number_of_orders,
    "total": total_sales,
    "mean": mean_sales
}

print("Report :", report)
````

## 3. From Python Lists to NumPy Arrays

Python lists are convenient, but data analysis often requires performing numerical operations on many values.

**NumPy** (Numerical Python) provides efficient multidimensional arrays and numerical operations.

Let's compare a Python list with a NumPy array.
**Main new content:** NumPy arrays, indexing, filtering, vectorization and descriptive statistics.

The sales examples describe small fictional business datasets. Their values and observation units are stated separately; they are not all the same dataset.


On crée la même série de ventes sous deux formes, une liste Python et un tableau NumPy, pour comparer leur comportement.

````python
import numpy as np

sales_list = [120, 135, 150, 165]

sales_array = np.array(sales_list)

print("Python list:", sales_list)
print("NumPy array:", sales_array)
````

À l'affichage, la liste a des virgules et le tableau NumPy n'en a pas, mais les valeurs sont les mêmes.

On multiplie la liste et le tableau par 2 pour voir la différence entre les deux structures.

````python
print("Python list:")
print(sales_list * 2)

print("\nNumPy array:")
print(sales_array * 2)
````

La liste est répétée deux fois, alors que le tableau NumPy multiplie chaque valeur par 2. C'est pour cette raison qu'on utilise NumPy pour les calculs.

### 3.1 Exploring NumPy arrays

A NumPy array has several useful properties:

- `shape` — dimensions of the array
- `size` — total number of elements
- `dtype` — data type of the elements
- `ndim` — number of dimensions

On regarde les propriétés d'un tableau NumPy, sa forme (`shape`), son nombre d'éléments (`size`), le type de ses valeurs (`dtype`) et son nombre de dimensions (`ndim`).

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("Array:", sales)
print("Shape:", sales.shape)
print("Size:", sales.size)
print("Data type:", sales.dtype)
print("Number of dimensions:", sales.ndim)
````

Le tableau contient 6 entiers sur une seule dimension, d'où la forme `(6,)`.

### 3.2 Indexing and slicing

NumPy arrays use the same basic indexing principles as Python lists.

We can access individual observations or select subsets of an array.

On sélectionne des valeurs du tableau par leur position, comme avec une liste. `[2:5]` prend les positions 2, 3 et 4.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("First value:", sales[0])
print("Last value:", sales[-1])
print("First three values:", sales[:3])
print("Values from index 2 to 4:", sales[2:5])
````

On obtient 120 en premier, 172 en dernier, et la tranche `[2:5]` donne bien trois valeurs, la borne de fin étant exclue.

### 3.3 Filtering data

NumPy can select observations that satisfy a condition.

The expression below creates a Boolean array containing `True` or `False` for each observation.

On compare chaque vente à 150. NumPy fait la comparaison sur tout le tableau d'un coup.

````python
sales >= 150
````

Le résultat est un masque booléen, avec `True` pour les ventes qui respectent la condition et `False` pour les autres.

On utilise ce masque entre crochets pour ne garder que les ventes supérieures ou égales à 150.

````python
print("Sales >= 150:", sales[sales >= 150])
````

Il reste 150, 165 et 172, c'est-à-dire les positions où le masque valait `True`.

### 3.4 Vectorized operations

NumPy allows us to perform an operation on an entire array without writing an explicit loop.

This is called **vectorization**.

On applique une hausse de 10 % à toutes les ventes en multipliant le tableau par 1,10. NumPy fait le calcul sur chaque valeur sans boucle, c'est la vectorisation.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

sales_after_increase = sales * 1.10

print("Original sales:", sales)
print("After 10% increase:", sales_after_increase)
````

Chaque vente est augmentée de 10 %. Les résultats deviennent décimaux, ce qui est normal après une multiplication par 1,10.

On soustrait les ventes d'origine aux ventes augmentées pour obtenir le montant de la hausse, encore une fois sur tout le tableau d'un coup.

````python
difference = sales_after_increase - sales

print("Increase:", difference)
````

La hausse représente bien 10 % de chaque vente, par exemple 12 pour une vente de 120.

### 3.5 Descriptive statistics

NumPy provides functions for computing common summary statistics.

These statistics help us obtain a first understanding of a dataset.

On calcule les statistiques descriptives principales avec NumPy, la moyenne, la médiane, le minimum, le maximum et l'écart-type.

````python
sales = np.array([120, 135, 128, 150, 165, 172])

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Standard deviation:", np.std(sales))
````

La moyenne (145) et la médiane (142,5) sont proches, donc il n'y a pas de valeur extrême qui tire la moyenne. L'écart-type d'environ 19 indique que les ventes s'écartent en moyenne d'une vingtaine d'euros de la moyenne.

**“Why might the median sometimes be more informative than the mean?”**

**Notre réponse.** La médiane peut être plus représentative lorsque les données contiennent des valeurs extrêmes, car elle est beaucoup moins influencée par ces valeurs que la moyenne.

Notre réponse à l'exercice sur les fonctions. On écrit la règle une seule fois dans `classify_sales`, puis une boucle l'applique à chaque vente.

````python
sales = [120, 95, 140, 180, 75, 160]

def classify_sales(value):
    if value >= 150:
        return "High"
    else:
        return "Standard"

for value in sales:
    print(value, "→", classify_sales(value))
````

**Pourquoi cette solution ?**  
Nous avons isolé la règle de classification dans une fonction afin d'éviter de répéter le même test. La boucle applique ensuite cette fonction à chaque valeur de la liste, ce qui rend la solution réutilisable et facile à modifier si le seuil change.
